```python
%%writefile /content/atr_colab_project/src/atr/hf_generator.py
"""Hugging Face generator for the Reflexion baseline."""

from typing import Optional


class HuggingFaceGenerator:
    """Callable text-generation wrapper around a causal language model."""

    def __init__(
        self,
        model_name: str = "Qwen/Qwen2.5-1.5B-Instruct",
        device: Optional[str] = None,
    ) -> None:
        import torch
        from transformers import AutoModelForCausalLM, AutoTokenizer

        self.torch = torch
        self.model_name = model_name

        if device is None:
            device = "cuda" if torch.cuda.is_available() else "cpu"

        self.device = torch.device(device)

        if self.device.type == "cuda":
            dtype = torch.float16
        else:
            dtype = torch.float32

        self.tokenizer = AutoTokenizer.from_pretrained(model_name)

        self.model = AutoModelForCausalLM.from_pretrained(
            model_name,
            torch_dtype=dtype,
        )

        self.model.to(self.device)
        self.model.eval()

        if self.tokenizer.pad_token_id is None:
            self.tokenizer.pad_token = self.tokenizer.eos_token

    def __call__(
        self,
        prompt: str,
        max_new_tokens: int = 128,
    ) -> str:
        if not prompt.strip():
            raise ValueError("prompt cannot be empty")

        if max_new_tokens < 1:
            raise ValueError("max_new_tokens must be >= 1")

        inputs = self.tokenizer(
            prompt,
            return_tensors="pt",
        )

        inputs = {
            key: value.to(self.device)
            for key, value in inputs.items()
        }

        with self.torch.inference_mode():
            output = self.model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                do_sample=False,
                pad_token_id=self.tokenizer.pad_token_id,
                eos_token_id=self.tokenizer.eos_token_id,
            )

        input_length = inputs["input_ids"].shape[1]
        new_tokens = output[0, input_length:]

        return self.tokenizer.decode(
            new_tokens,
            skip_special_tokens=True,
        ).strip()
```

```python
%%writefile /content/atr_colab_project/scripts/test_workflow_loop.py
import os
import sys

PROJECT_ROOT = "/content/atr_colab_project"
SRC_DIR = os.path.join(PROJECT_ROOT, "src")

if SRC_DIR not in sys.path:
    sys.path.insert(0, SRC_DIR)

from atr.hf_generator import HuggingFaceGenerator
from atr.reflexion_agent import ReflexionAgent
from atr.workflow_env import ResearchWorkflowEnv


def main():
    env = ResearchWorkflowEnv()
    state_summary = env.reset()

    generator = HuggingFaceGenerator(
        model_name="Qwen/Qwen2.5-1.5B-Instruct"
    )

    agent = ReflexionAgent(
        generate_text=generator,
        max_new_tokens=64,
    )

    task_id = "workflow_smoke_test"
    task_instruction = (
        "Prepare a research dataset. "
        "Available actions: download_dataset, check_schema, "
        "validate_dataset. Perform the next required action."
    )

    action_text = agent.act(
        task_id=task_id,
        task_instruction=task_instruction,
        current_state_summary=state_summary,
    )

    # Convert the LLM proposal into one of the environment's fixed actions.
    # This simple mapping is for the smoke test, not the final action parser.
    text = action_text.lower()

    if "download" in text:
        action = "download_dataset"
    elif "schema" in text:
        action = "check_schema"
    elif "validat" in text:
        action = "validate_dataset"
    else:
        action = "unknown"

    feedback, done = env.step(action)

    reflection = agent.reflect(
        task_id=task_id,
        task_instruction=task_instruction,
        attempted_action=action,
        environment_feedback=feedback,
        trajectory_summary=(
            f"Initial state: {state_summary}\n"
            f"Proposed action: {action_text}"
        ),
        trial=1,
    )

    assert isinstance(feedback, str) and feedback
    assert isinstance(reflection, str) and reflection
    assert len(agent.memory.recall(task_id)) == 1

    print("WORKFLOW_LOOP_TEST: PASS")
    print("proposed_action:", action_text)
    print("mapped_action:", action)
    print("environment_feedback:", feedback)
    print("done:", done)
    print("reflection_characters:", len(reflection))
    print("stored_reflections:", len(agent.memory.recall(task_id)))


if __name__ == "__main__":
    main()
```

```python
%%writefile /content/atr_colab_project/src/atr/belief_state.py
"""Agent belief state, kept separate from the true environment state."""

from copy import deepcopy


class BeliefState:
    def __init__(self, initial_state):
        self.state = deepcopy(initial_state)

    def get(self, key):
        return self.state[key]

    def set(self, key, value):
        if key not in self.state:
            raise KeyError(f"Unknown belief field: {key}")

        self.state[key] = value

    def snapshot(self):
        return deepcopy(self.state)

    def summary(self):
        return "; ".join(
            f"{key}={value}"
            for key, value in self.state.items()
        )
```

```python
%%writefile /content/atr_colab_project/src/atr/fault_injector.py
"""Controlled fault injection into agent belief state."""

from copy import deepcopy


class FaultInjector:
    def inject_boolean_error(
        self,
        belief,
        field,
        true_state,
        error_id="injected_error_1",
    ):
        if field not in true_state:
            raise KeyError(f"Unknown true-state field: {field}")

        if field not in belief.state:
            raise KeyError(f"Unknown belief field: {field}")

        true_value = true_state[field]

        if not isinstance(true_value, bool):
            raise TypeError("Fault injection currently supports Boolean fields only")

        # Record the error for evaluation, not for the agent.
        record = {
            "error_id": error_id,
            "field": field,
            "true_value": true_value,
            "injected_value": not true_value,
        }

        # Change only the agent's belief.
        belief.set(field, not true_value)

        return record
```

```python
%%writefile /content/atr_colab_project/scripts/test_fault_injection.py
import os
import sys

PROJECT_ROOT = "/content/atr_colab_project"
SRC_DIR = os.path.join(PROJECT_ROOT, "src")

if SRC_DIR not in sys.path:
    sys.path.insert(0, SRC_DIR)

from atr.belief_state import BeliefState
from atr.fault_injector import FaultInjector


def main():
    true_state = {
        "dataset_downloaded": True,
        "schema_checked": True,
        "dataset_validated": False,
    }

    true_before = true_state.copy()
    belief = BeliefState(true_state)

    injector = FaultInjector()

    record = injector.inject_boolean_error(
        belief=belief,
        field="dataset_validated",
        true_state=true_state,
        error_id="test_error_1",
    )

    # Belief must now be wrong for the selected field.
    assert belief.get("dataset_validated") is True

    # The authoritative environment state must not change.
    assert true_state == true_before
    assert true_state["dataset_validated"] is False

    # Other belief fields must remain unchanged.
    assert belief.get("dataset_downloaded") is True
    assert belief.get("schema_checked") is True

    # Evaluation record should describe the injected mismatch.
    assert record["true_value"] is False
    assert record["injected_value"] is True

    print("FAULT_INJECTION_TEST: PASS")
    print("true_state_unchanged:", true_state == true_before)
    print("belief_dataset_validated:", belief.get("dataset_validated"))
    print("true_dataset_validated:", true_state["dataset_validated"])
    print("error_record:", record)


if __name__ == "__main__":
    main()
```

```python
%%writefile /content/atr_colab_project/scripts/test_baseline_episode.py
import os
import sys
import re

ROOT = "/content/atr_colab_project"
SRC_DIR = os.path.join(ROOT, "src")

if SRC_DIR not in sys.path:
    sys.path.insert(0, SRC_DIR)

from atr.hf_generator import HuggingFaceGenerator
from atr.reflexion_agent import ReflexionAgent
from atr.workflow_env import ResearchWorkflowEnv
from atr.belief_state import BeliefState
from atr.fault_injector import FaultInjector


def parse_action(text):
    allowed = [
        "download_dataset",
        "check_schema",
        "validate_dataset",
    ]

    text = text.lower()

    # First, look for an explicit action label.
    match = re.search(
        r"(?:next action|action)\s*:\s*"
        r"(download_dataset|check_schema|validate_dataset)\b",
        text,
    )

    if match:
        return match.group(1)

    # Otherwise, accept exactly one valid action identifier.
    matches = [
        action
        for action in allowed
        if re.search(rf"\b{re.escape(action)}\b", text)
    ]

    if len(matches) == 1:
        return matches[0]

    return "invalid_action"


def main():
    # 1. Create the environment.
    env = ResearchWorkflowEnv()
    env.reset()

    # Set prerequisite state for this episode.
    env.state["dataset_downloaded"] = True
    env.state["schema_checked"] = True
    env.state["dataset_validated"] = False

    true_before_injection = env.state.copy()

    # 2. Create the agent's separate belief.
    belief = BeliefState(env.state)

    # 3. Inject a false belief without changing true state.
    injector = FaultInjector()

    error_record = injector.inject_boolean_error(
        belief=belief,
        field="dataset_validated",
        true_state=env.state,
        error_id="b2_episode_error_1",
    )

    assert env.state == true_before_injection
    assert belief.get("dataset_validated") is True

    # 4. Load the model and Reflexion-style agent.
    generator = HuggingFaceGenerator(
        model_name="Qwen/Qwen2.5-1.5B-Instruct"
    )

    agent = ReflexionAgent(
        generate_text=generator,
        max_new_tokens=64,
    )

    task_id = "b2_baseline_smoke_test"

    task = (
        "Prepare the dataset. Choose exactly one next action from: "
        "download_dataset, check_schema, validate_dataset. "
        "Return the action identifier only."
    )

    # 5. Actor proposes an action using its belief.
    proposal = agent.act(
        task_id=task_id,
        task_instruction=task,
        current_state_summary=belief.summary(),
    )

    # 6. Parse the proposal.
    action = parse_action(proposal)

    # 7. Execute the selected action in the environment.
    feedback, done = env.step(action)

    # 8. Reflect on the actual environment feedback.
    reflection = agent.reflect(
        task_id=task_id,
        task_instruction=task,
        attempted_action=action,
        environment_feedback=feedback,
        trajectory_summary=(
            f"Belief before action: {belief.summary()}\n"
            f"Actor proposal: {proposal}\n"
            f"Parsed action: {action}"
        ),
        trial=1,
    )

    # 9. Print diagnostic output.
    print("B2_BASELINE_EPISODE_TEST: COMPLETE")
    print("true_state_unchanged_by_injection:",
          env.state == true_before_injection)
    print("injected_field:", error_record["field"])
    print("actor_proposal:", proposal.replace("\n", " ")[:300])
    print("parsed_action:", action)
    print("environment_feedback:", feedback)
    print("task_done:", done)
    print("true_validation_state:", env.state["dataset_validated"])
    print("reflection_characters:", len(reflection))
    print("stored_reflections:", len(agent.memory.recall(task_id)))


if __name__ == "__main__":
    main()
```

```python
%%writefile /content/atr_colab_project/src/atr/hf_generator.py
"""Qwen chat-template generator for the ATR Reflexion-style baseline."""

from typing import Optional


class HuggingFaceGenerator:
    """Callable text-generation wrapper around a causal language model."""

    def __init__(
        self,
        model_name: str = "Qwen/Qwen2.5-1.5B-Instruct",
        device: Optional[str] = None,
    ) -> None:
        import torch
        from transformers import AutoModelForCausalLM, AutoTokenizer

        self.torch = torch
        self.model_name = model_name

        if device is None:
            device = "cuda" if torch.cuda.is_available() else "cpu"

        self.device = torch.device(device)
        dtype = torch.float16 if self.device.type == "cuda" else torch.float32

        self.tokenizer = AutoTokenizer.from_pretrained(model_name)

        self.model = AutoModelForCausalLM.from_pretrained(
            model_name,
            torch_dtype=dtype,
        )

        self.model.to(self.device)
        self.model.eval()

        if self.tokenizer.pad_token_id is None:
            self.tokenizer.pad_token = self.tokenizer.eos_token

    def __call__(
        self,
        prompt: str,
        max_new_tokens: int = 128,
    ) -> str:
        if not prompt.strip():
            raise ValueError("prompt cannot be empty")

        if max_new_tokens < 1:
            raise ValueError("max_new_tokens must be >= 1")

        # Convert the existing prompt into a user message.
        messages = [
            {
                "role": "user",
                "content": prompt,
            }
        ]

        # Use the model's official chat template when available.
        if getattr(self.tokenizer, "chat_template", None):
            inputs = self.tokenizer.apply_chat_template(
                messages,
                tokenize=True,
                add_generation_prompt=True,
                return_tensors="pt",
                return_dict=True,
            )
        else:
            # Fallback for tokenizers without a chat template.
            inputs = self.tokenizer(
                prompt,
                return_tensors="pt",
            )

        inputs = {
            key: value.to(self.device)
            for key, value in inputs.items()
        }

        input_length = inputs["input_ids"].shape[1]

        with self.torch.inference_mode():
            output = self.model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                do_sample=False,
                pad_token_id=self.tokenizer.pad_token_id,
                eos_token_id=self.tokenizer.eos_token_id,
            )

        new_tokens = output[0, input_length:]

        response = self.tokenizer.decode(
            new_tokens,
            skip_special_tokens=True,
        ).strip()

        return response
```

```python
%%writefile /content/atr_colab_project/scripts/test_reflection_retry.py
import os
import sys
import re

ROOT = "/content/atr_colab_project"
sys.path.insert(0, os.path.join(ROOT, "src"))

from atr.hf_generator import HuggingFaceGenerator
from atr.reflexion_agent import ReflexionAgent
from atr.workflow_env import ResearchWorkflowEnv
from atr.belief_state import BeliefState
from atr.fault_injector import FaultInjector


def parse_action(text):
    allowed = {
        "download_dataset",
        "check_schema",
        "validate_dataset",
    }

    text = text.lower()

    # Prefer an explicitly labelled action.
    match = re.search(
        r"(?:next action|action)\s*:\s*"
        r"(download_dataset|check_schema|validate_dataset)\b",
        text,
    )

    if match:
        return match.group(1)

    # Otherwise accept only one unique allowed action.
    matches = [
        action for action in allowed
        if re.search(rf"\b{re.escape(action)}\b", text)
    ]

    return matches[0] if len(matches) == 1 else "invalid_action"


def main():
    env = ResearchWorkflowEnv()
    env.reset()

    # Prepare the workflow state for this test.
    env.state["dataset_downloaded"] = True
    env.state["schema_checked"] = True
    env.state["dataset_validated"] = False

    true_state_before_injection = env.state.copy()

    belief = BeliefState(env.state)

    injector = FaultInjector()
    error_record = injector.inject_boolean_error(
        belief=belief,
        field="dataset_validated",
        true_state=env.state,
        error_id="retry_test_error_1",
    )

    assert env.state == true_state_before_injection

    generator = HuggingFaceGenerator()
    agent = ReflexionAgent(
        generate_text=generator,
        max_new_tokens=64,
    )

    task_id = "reflection_retry_test"

    task = (
        "Prepare the dataset. Choose exactly one next action from: "
        "download_dataset, check_schema, validate_dataset. "
        "Return only the action identifier."
    )

    # Attempt 1.
    proposal_1 = agent.act(
        task_id=task_id,
        task_instruction=task,
        current_state_summary=belief.summary(),
    )

    action_1 = parse_action(proposal_1)
    feedback_1, done_1 = env.step(action_1)

    # Reflection on the first attempt.
    reflection_1 = agent.reflect(
        task_id=task_id,
        task_instruction=task,
        attempted_action=action_1,
        environment_feedback=feedback_1,
        trajectory_summary=(
            f"Belief: {belief.summary()}\n"
            f"Proposal: {proposal_1}\n"
            f"Action: {action_1}\n"
            f"Feedback: {feedback_1}"
        ),
        trial=1,
    )

    # Attempt 2: expose prior feedback and reflection to the agent.
    retry_task = (
        task
        + "\nPrevious action: " + action_1
        + "\nPrevious environment feedback: " + feedback_1
        + "\nYour reflection: " + reflection_1
        + "\nChoose the next valid action. Return only its identifier."
    )

    proposal_2 = agent.act(
        task_id=task_id,
        task_instruction=retry_task,
        current_state_summary=belief.summary(),
    )

    action_2 = parse_action(proposal_2)
    feedback_2, done_2 = env.step(action_2)

    print("REFLECTION_RETRY_TEST: COMPLETE")
    print("injected_field:", error_record["field"])
    print("true_state_unchanged_by_injection:",
          env.state["dataset_validated"] is False)
    print("attempt_1_proposal:", proposal_1.replace("\n", " ")[:200])
    print("attempt_1_action:", action_1)
    print("attempt_1_feedback:", feedback_1)
    print("reflection_characters:", len(reflection_1))
    print("attempt_2_proposal:", proposal_2.replace("\n", " ")[:200])
    print("attempt_2_action:", action_2)
    print("attempt_2_feedback:", feedback_2)
    print("task_done_after_retry:", done_2)
    print("true_validation_state:", env.state["dataset_validated"])


if __name__ == "__main__":
    main()
```

```python
# ATR PILOT EXPERIMENT — ONE COLAB CELL
# Conditions: B1 Direct | B2 Reflection | P Reflection + verifier report
# Horizons: 4, 8, 12
# Outputs: episode_results.csv, aggregate_results.csv, recovery_plot.png

import os
import json
import time
import random
import pandas as pd
import matplotlib.pyplot as plt
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# ---------------- CONFIG ----------------
SEED = 20261030
MODEL_NAME = "Qwen/Qwen2.5-1.5B-Instruct"
HORIZONS = [4, 8, 12]
CONDITIONS = ["B1", "B2", "P"]
TRIALS_PER_SETTING = 2
MAX_NEW_TOKENS = 80
OUTPUT_DIR = "/content/atr_results"

random.seed(SEED)
torch.manual_seed(SEED)
os.makedirs(OUTPUT_DIR, exist_ok=True)

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", DEVICE)

# ---------------- LOAD MODEL ----------------
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16 if DEVICE == "cuda" else torch.float32,
    device_map="auto" if DEVICE == "cuda" else None
)
model.eval()

# ---------------- GENERATION ----------------
def generate_text(prompt):
    messages = [{"role": "user", "content": prompt}]

    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt",
        return_dict=True
    )

    inputs = {
        key: value.to(model.device)
        for key, value in inputs.items()
    }

    input_length = inputs["input_ids"].shape[1]

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=MAX_NEW_TOKENS,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    return tokenizer.decode(
        output[0][input_length:],
        skip_special_tokens=True
    ).strip()

# ---------------- WORKFLOW ----------------
ACTIONS = [
    "identify_research_question",
    "define_scope",
    "search_sources",
    "screen_sources",
    "extract_evidence",
    "check_evidence_quality",
    "organize_findings",
    "analyze_findings",
    "identify_gaps",
    "draft_synthesis",
    "check_citations",
    "finalize_report"
]

DEPENDENCIES = {
    ACTIONS[0]: [],
    ACTIONS[1]: [ACTIONS[0]],
    ACTIONS[2]: [ACTIONS[1]],
    ACTIONS[3]: [ACTIONS[2]],
    ACTIONS[4]: [ACTIONS[3]],
    ACTIONS[5]: [ACTIONS[4]],
    ACTIONS[6]: [ACTIONS[5]],
    ACTIONS[7]: [ACTIONS[6]],
    ACTIONS[8]: [ACTIONS[7]],
    ACTIONS[9]: [ACTIONS[8]],
    ACTIONS[10]: [ACTIONS[9]],
    ACTIONS[11]: [ACTIONS[10]]
}

def initial_state():
    return {a: False for a in ACTIONS}

def valid_actions(state, horizon):
    return [
        a for a in ACTIONS[:horizon]
        if not state[a]
        and all(state[p] for p in DEPENDENCIES[a])
    ]

def execute_action(state, action):
    if action not in ACTIONS:
        return "ERROR: unknown action", False
    if state[action]:
        return "ERROR: action already completed", False
    if not all(state[p] for p in DEPENDENCIES[action]):
        return "ERROR: prerequisites not completed", False

    state[action] = True
    return f"Completed: {action}", True

def task_complete(state, horizon):
    return all(state[a] for a in ACTIONS[:horizon])

# ---------------- ACTION SELECTION ----------------
def choose_action(belief, horizon, reflection="", verifier_report=""):
    candidates = valid_actions(belief, horizon)

    if not candidates:
        return None, ""

    prompt = f"""
You are completing a research workflow.

Belief state:
{json.dumps(belief)}

Available actions:
{candidates}

Reflection:
{reflection}

Verifier report:
{verifier_report}

Choose exactly one action from Available actions.
Return only its exact identifier.
"""

    raw = generate_text(prompt)

    # Exact action matching; no forced fallback.
    matches = [
        candidate for candidate in candidates
        if candidate in raw
    ]

    return (matches[0] if len(matches) == 1 else None), raw

# ---------------- EPISODE RUNNER ----------------
def run_episode(condition, horizon, trial, injected):
    true_state = initial_state()
    belief = initial_state()

    reflection = ""
    logs = []
    injected_field = ACTIONS[min(1, horizon - 1)]
    injection_done = False
    recovered = False
    start_time = time.time()

    for step in range(horizon + 4):

        # Inject an error into belief only.
        if injected and not injection_done and step == 1:
            belief[injected_field] = not true_state[injected_field]
            injection_done = True

        # Independent state-consistency report.
        verifier_report = ""
        if condition == "P" and injection_done:
            mismatches = [
                key for key in belief
                if belief[key] != true_state[key]
            ]
            if mismatches:
                verifier_report = (
                    "Mismatch detected for: " + ", ".join(mismatches)
                )

        action, raw = choose_action(
            belief,
            horizon,
            reflection if condition in ["B2", "P"] else "",
            verifier_report
        )

        if action is None:
            feedback = "ERROR: invalid or ambiguous action output"
            valid = False
        else:
            feedback, valid = execute_action(true_state, action)

        # Update belief only from successful action feedback.
        if valid:
            belief[action] = True

        # Reflection condition.
        if condition in ["B2", "P"]:
            reflection = generate_text(
                f"""
Action: {action}
Environment feedback: {feedback}
Belief state: {json.dumps(belief)}
Available actions: {valid_actions(belief, horizon)}
Verifier report: {verifier_report}

Reflect briefly. State the next action or correction.
"""
            )

        # Recovery means the injected field matches true state.
        if injection_done and belief[injected_field] == true_state[injected_field]:
            recovered = True

        logs.append({
            "step": step,
            "action": action,
            "raw_output": raw,
            "feedback": feedback,
            "valid": valid,
            "reflection": reflection,
            "verifier_report": verifier_report
        })

        if task_complete(true_state, horizon):
            break

    return {
        "condition": condition,
        "horizon": horizon,
        "trial": trial,
        "injected": injected,
        "success": task_complete(true_state, horizon),
        "recovered": recovered if injected else None,
        "steps": len(logs),
        "runtime_seconds": time.time() - start_time,
        "trajectory": json.dumps(logs)
    }

# ---------------- RUN EXPERIMENT ----------------
results = []
total = len(CONDITIONS) * len(HORIZONS) * TRIALS_PER_SETTING * 2
count = 0

for horizon in HORIZONS:
    for condition in CONDITIONS:
        for injected in [False, True]:
            for trial in range(TRIALS_PER_SETTING):
                count += 1
                print(
                    f"{count}/{total} | {condition} | "
                    f"horizon={horizon} | injected={injected} | "
                    f"trial={trial}"
                )

                results.append(
                    run_episode(condition, horizon, trial, injected)
                )

# ---------------- SAVE EPISODES ----------------
results_df = pd.DataFrame(results)

episode_path = os.path.join(OUTPUT_DIR, "episode_results.csv")
results_df.to_csv(episode_path, index=False)

# ---------------- AGGREGATE ----------------
aggregate = (
    results_df.groupby(["condition", "horizon", "injected"])
    .agg(
        episodes=("success", "count"),
        success_rate=("success", "mean"),
        recovery_rate=("recovered", "mean"),
        mean_steps=("steps", "mean"),
        mean_runtime_seconds=("runtime_seconds", "mean")
    )
    .reset_index()
)

aggregate_path = os.path.join(OUTPUT_DIR, "aggregate_results.csv")
aggregate.to_csv(aggregate_path, index=False)

# ---------------- PLOT ----------------
injected_results = aggregate[aggregate["injected"] == True]

plt.figure(figsize=(8, 5))

for condition in CONDITIONS:
    subset = injected_results[
        injected_results["condition"] == condition
    ]

    plt.plot(
        subset["horizon"],
        subset["recovery_rate"],
        marker="o",
        label=condition
    )

plt.xlabel("Workflow horizon")
plt.ylabel("Injected-error recovery rate")
plt.title("Recovery rate by condition and horizon")
plt.xticks(HORIZONS)
plt.ylim(0, 1)
plt.grid(True)
plt.legend()

plot_path = os.path.join(OUTPUT_DIR, "recovery_plot.png")
plt.savefig(plot_path, dpi=150, bbox_inches="tight")
plt.show()

print("\nPILOT RUN FINISHED")
print("Episode CSV:", episode_path)
print("Aggregate CSV:", aggregate_path)
print("Plot:", plot_path)

display(aggregate)
```
